## 2.1 Dasar Teori

Praktikum Pengenalan Python, VS Code, dan Koneksi ke GitHub membahas dasar penggunaan bahasa pemrograman Python, Visual Studio Code sebagai code editor, serta GitHub sebagai media penyimpanan dan pengelolaan kode program.

Praktikum Tipe Data dan Variabel membahas konsep dasar dalam bahasa pemrograman Python yang digunakan untuk menyimpan dan mengolah data. Tipe data menentukan jenis nilai yang digunakan dalam program, sedangkan variabel digunakan untuk menyimpan atau mereferensikan nilai tersebut.

1. Integer (int)<br>
Integer merupakan tipe data yang digunakan untuk menyimpan bilangan bulat, seperti 10, 25, dan -5.
2. Float (float)<br>
Float digunakan untuk menyimpan bilangan yang memiliki nilai desimal, seperti 3.14, 10.5, dan 0.75.
3. Complex (complex)<br>
Complex merupakan tipe data yang digunakan untuk menyimpan bilangan kompleks yang memiliki bagian real dan imajiner. Contohnya adalah 3+2j.
4. String (str)<br>
String merupakan tipe data yang digunakan untuk menyimpan teks atau kumpulan karakter. String ditulis menggunakan tanda kutip, contohnya "Python" dan "Informatika".
5. Boolean (bool)<br>
Boolean merupakan tipe data yang memiliki dua nilai, yaitu True dan False. Tipe data ini biasanya digunakan untuk menyatakan kondisi benar atau salah.
6. Konstanta<br>
Konstanta merupakan nilai yang secara konsep tidak dimaksudkan untuk diubah selama program berjalan. Python tidak memiliki sintaks khusus untuk membuat konstanta, tetapi penamaan dengan huruf kapital dapat digunakan sebagai tanda bahwa suatu variabel dianggap sebagai konstanta, contohnya PI = 3.14.
7. Variabel<br>
Variabel digunakan sebagai tempat untuk menyimpan atau mereferensikan suatu nilai. Dalam Python, variabel dapat dibuat dengan memberikan nilai menggunakan operator =, misalnya nama = "Zaky" atau umur = 20.
8. Input dan Output<br>
Python menyediakan fungsi input() untuk menerima data dari pengguna dan print() untuk menampilkan data ke layar. Data dari input() secara default bertipe str, sehingga perlu dilakukan konversi apabila data tersebut akan digunakan sebagai bilangan.

### 2.2.5 Program Konstanta_2611533029
</div>

```python
# Buat file dengan nama Konstanta_2611533029
# Program ini menggunakan konstanta untuk menghitun luas lingkaran
# nama variabel ditambah 4 digit nim terakhir contoh: jari_3029

from typing import Final
PI: Final = 3.14
print("Pi: %f" % (PI))
jari_3029 = float(input("Masukkan nilai jari-jari: "))
luas_3029 = PI * jari_3029
print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_3029,luas_3029))
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

Penjelasan:

1. Pemanggilan Final dari library typing from typing import Final
2. Deklarasi PI: Final = 3.14 sebagai konstanta
3. Cetak ("Pi: %f" % (PI))
4. Nilai jari_3029 = float(input("Masukkan nilai jari-jari: "))
5. Nilai luas_3029 = PI * jari_3029
6. Cetak ("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_3029,luas_3029))
