### 2.1.1
Perulangan for merupakan salah satu struktur perulangan yang tersedia dalam Python. Perulangan ini digunakan untuk mengakses setiap elemen dari suatu urutan atau iterable, seperti list, tuple, string, dan range. Setiap elemen akan diproses secara berurutan hingga seluruh elemen dalam iterable selesai diproses.

### 2.1.2
range() merupakan fungsi yang sering digunakan bersama perulangan for untuk menghasilkan urutan bilangan. Fungsi ini dapat digunakan untuk menentukan nilai awal, batas akhir, dan jarak perubahan nilai dalam suatu perulangan.

### 2.1.3
Nested for adalah perulangan for yang berada di dalam perulangan for lainnya. Struktur ini biasanya digunakan ketika program membutuhkan perulangan bertingkat, misalnya untuk mengolah data dalam bentuk baris dan kolom atau membuat pola tertentu.

### 2.2.1 Program perulangan_for1_2611533029

```python
# Buat file dengan nama perulangan_for1_NIM.py
# Buat program untuk perulangan for dalam pytohn
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_3029 = int(input("Masukkan jumlah perulangan: "))

for i_3029 in range(ulang_3029):
    print(f"Perulangan ke-{i_3029}")
```

5. `print()` berfungsi memindahkan output ke baris baru setelah perulangan pertama selesai.
6. `print("perulangan ke-1 sampai ke-", ulang_3029)` berfungsi menampilkan keterangan bahwa perulangan berikutnya akan mencetak angka dari `1` sampai nilai `ulang_3029`.

### 2.2.3 Program perulangan_for3_2611533029
</div>

```python
# Buat file dengan nama perulangan_for3_NIM.py
# Buat program untuk perulangan for dalam pytohn
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_3029 = int(input("Masukkan jumlah perulangan: "))

jumlah_3029 = 0
for i_3029 in range(1,ulang_3029+1):
    print(i_3029,end=" ")
    jumlah_3029 = jumlah_3029 + i_3029

    if i_3029 < ulang_3029:
        print(" + ",end=" ")
    else:
        print(" = ",jumlah_3029,end=" ")
print()
print("Jumlah =",jumlah_3029)
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

Penjelasan:
1. lang_3029 = int(input("Masukkan jumlah perulangan: ")) berfungsi untuk mengambil jumlah perulangan dari pengguna, lalu mengonversi masukan tersebut menjadi bilangan bulat (integer).
2. jumlah_3029 = 0 berfungsi untuk membuat variabel yang menampung total penjumlahan, dengan nilai awal 0.
3. for i_3029 in range(1, ulang_3029+1): berfungsi untuk mengulang proses dari angka 1 sampai ulang_3029. Batas akhir range() ditambah 1 supaya nilai ulang_3029 juga ikut diproses.
4. print(i_3029, end=" ") berfungsi untuk mencetak setiap angka dalam satu baris yang sama. Parameter end=" " menambahkan spasi setelah tiap angka.
5. jumlah_3029 = jumlah_3029 + i_3029 berfungsi untuk menambahkan nilai i_3029 ke jumlah_3029, sehingga semua angka yang ditampilkan ikut terjumlahkan.
6. if i_3029 < ulang_3029: berfungsi untuk mengecek apakah angka yang sedang diproses masih lebih kecil daripada jumlah perulangan.
7. print(" + ", end=" ") berfungsi untuk mencetak tanda + di antara angka-angka yang dijumlahkan.
else: akan dieksekusi saat i_3029 sudah sampai pada angka terakhir dalam perulangan.
8. print(" = ", jumlah_3029, end=" ") berfungsi untuk mencetak tanda = bersama hasil penjumlahan seluruh angka.
9. print() berfungsi untuk pindah ke baris baru setelah perulangan selesai.
print("Jumlah =", jumlah_3029) berfungsi untuk menampilkan hasil akhir penjumlahan yang tersimpan dalam variabel jumlah_3029.


### 2.2.4 Program nested_for1_2611533029
</div>

```python
# Buat file dengan nama nested_for1_NIM.py
# Buat program untuk perulangan for dalam pytohn
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

batas_3029 = int(input("Masukkan nilai batas: "))
for line_3029 in range(1,batas_3029 + 1):
    for j_3029 in range(1,(-1 * line_3029 + batas_3029) + 1):
        print(".",end="")
    print(line_3029)
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

Penjelasan:

2. `for line_3029 in range(1, batas_3029 + 1):` berfungsi sebagai perulangan luar yang berjalan dari angka `1` sampai `batas_3029`.
3. `for j_3029 in range(1, (-1 * line_3029 + batas_3029) + 1):` berfungsi sebagai perulangan dalam yang menentukan banyaknya karakter titik (`.`) yang dicetak. Semakin besar nilai `line_3029`, semakin sedikit titik yang muncul.
4. `print(".", end="")` berfungsi untuk mencetak titik tanpa pindah ke baris baru, sehingga titik-titik tercetak berurutan dalam satu baris. Hal ini dicapai lewat parameter `end=""`.
5. `print(line_3029)` berfungsi untuk menampilkan nilai `line_3029` setelah semua titik pada baris tersebut selesai dicetak.

### 2.2.7 Program nested_for4_2611533029
</div>

```python
# Buat file dengan nama nested_for4_NIM.py
# Buat program untuk perulangan for dalam pytohn
# Nama variabel ditambah 4 digit nim terakhir c_3029ontoh: ulang_1234
# Program ini menggunakan fungsi input()

tinggi_3029 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_3029 % 2 != 0:
    print("Tinggi pola harus bilangan genap.")
else:
    a_3029 = tinggi_3029
    c_3029 = a_3029
    lebar_3029 = (2 * tinggi_3029) - 2

    for i_3029 in range(1, tinggi_3029 + 1):
        b_3029 = c_3029 + 1

        for j_3029 in range(1, lebar_3029 + 1):

            # Baris atas dan bawah
            if i_3029 == 1 or i_3029 == tinggi_3029:
                if j_3029 == 1 or j_3029 == lebar_3029:
                    print("#", end="")
                else:
                    print("*", end="")
            # Baris isi
            else:
                if j_3029 == 1 or j_3029 == lebar_3029:
                    print("|", end="")
                else:
                    if j_3029 == c_3029:
                        print("<", end="")
                    elif j_3029 == b_3029:
                        print(">", end="")
                    elif j_3029 == (lebar_3029 - c_3029):
                        print("<", end="")
                    elif j_3029 == (lebar_3029 - c_3029 + 1):
                        print(">", end="")
                    elif j_3029 > b_3029 and j_3029 < (lebar_3029 - c_3029):
                        print(".", end="")
                    else:
                        print(" ", end="")

        print()

        # logika asli java
        a_3029 -= 2
        if a_3029 <= 0:
            c_3029 = -a_3029 + 2
        else:
            c_3029 = a_3029
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

Penjelasan:
1. `tinggi_3029 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))` berfungsi untuk meminta pengguna memasukkan tinggi pola, lalu mengonversi masukan tersebut menjadi bilangan bulat (integer).

4. `else:` berfungsi untuk menjalankan kode pembentuk pola jika `tinggi_3029` adalah bilangan genap.
5. `a_3029 = tinggi_3029` dan `c_3029 = a_3029` berfungsi untuk menyimpan nilai awal yang nantinya dipakai untuk menentukan posisi karakter dalam pola.
6. `lebar_3029 = (2 * tinggi_3029) - 2` berfungsi untuk menghitung lebar pola berdasarkan tinggi yang dimasukkan.
7. `for i_3029 in range(1, tinggi_3029 + 1):` berfungsi untuk mengulang proses pada setiap baris pola, dari baris pertama sampai baris terakhir.
8. `b_3029 = c_3029 + 1` berfungsi untuk menentukan posisi karakter `>` berdasarkan nilai `c_3029`.
9. `for j_3029 in range(1, lebar_3029 + 1):` berfungsi untuk mengulang proses pada setiap posisi kolom dalam satu baris.
10. `if i_3029 == 1 or i_3029 == tinggi_3029:` berfungsi untuk mengecek apakah baris yang sedang diproses adalah baris paling atas atau paling bawah.
11. `if j_3029 == 1 or j_3029 == lebar_3029:` berfungsi untuk menentukan posisi sudut kiri dan kanan pada baris atas atau bawah. Pada posisi itu program mencetak karakter `#`.
12. `print("*", end="")` berfungsi untuk mencetak karakter `*` pada bagian selain sudut di baris paling atas dan bawah.
13. `if j_3029 == 1 or j_3029 == lebar_3029:` pada bagian baris isi berfungsi untuk mencetak karakter `|` di sisi kiri dan kanan pola.
14. `if j_3029 == c_3029:` berfungsi untuk menentukan posisi karakter `<` pada bagian dalam pola.
15. `elif j_3029 == b_3029:` berfungsi untuk menentukan posisi karakter `>` berdasarkan nilai `b_3029`.
16. `elif j_3029 == (lebar_3029 - c_3029):` dan `elif j_3029 == (lebar_3029 - c_3029 + 1):` berfungsi untuk menentukan posisi karakter `<` dan `>` di sisi pola yang berlawanan.
17. `elif j_3029 > b_3029 and j_3029 < (lebar_3029 - c_3029):` berfungsi untuk menentukan ruang di antara posisi karakter tertentu. Pada kondisi ini program mencetak karakter `.`.
18. `print(" ", end="")` berfungsi untuk mencetak spasi pada posisi yang tidak memenuhi kondisi sebelumnya, sehingga pola tampil pada posisi yang sesuai.
19. `print()` berfungsi untuk pindah ke baris berikutnya setelah semua kolom pada satu baris selesai diproses.
20. `a_3029 -= 2` berfungsi untuk mengurangi nilai `a_3029` sebesar `2` setiap kali satu baris selesai diproses.
21. `if a_3029 <= 0:` berfungsi untuk mengecek apakah nilai `a_3029` sudah mencapai atau melewati `0`.
22. `c_3029 = -a_3029 + 2` berfungsi untuk menghitung ulang nilai `c_3029` ketika `a_3029` sudah mencapai batas tersebut.
23. `else: c_3029 = a_3029` berfungsi untuk menetapkan `c_3029` sesuai nilai `a_3029` selama nilainya masih lebih besar dari `0`.
