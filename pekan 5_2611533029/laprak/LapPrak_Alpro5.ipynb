## 2.1 Dasar Teori
Perulangan merupakan salah satu konsep dasar dalam pemrograman yang digunakan untuk menjalankan suatu blok perintah secara berulang. Penggunaan perulangan dapat mengurangi penulisan kode yang sama secara berulang sehingga program menjadi lebih singkat dan efisien. Perulangan biasanya digunakan ketika suatu proses harus dilakukan beberapa kali berdasarkan jumlah tertentu atau kondisi yang telah ditentukan.

### 2.1.1
Perulangan for merupakan salah satu struktur perulangan yang tersedia dalam Python. Perulangan ini digunakan untuk mengakses setiap elemen dari suatu urutan atau iterable, seperti list, tuple, string, dan range. Setiap elemen akan diproses secara berurutan hingga seluruh elemen dalam iterable selesai diproses.

### 2.1.2
range() merupakan fungsi yang sering digunakan bersama perulangan for untuk menghasilkan urutan bilangan. Fungsi ini dapat digunakan untuk menentukan nilai awal, batas akhir, dan jarak perubahan nilai dalam suatu perulangan.

### 2.1.3
Nested for adalah perulangan for yang berada di dalam perulangan for lainnya. Struktur ini biasanya digunakan ketika program membutuhkan perulangan bertingkat, misalnya untuk mengolah data dalam bentuk baris dan kolom atau membuat pola tertentu.

## 2.2 Kode Program

### 2.2.1 Program perulangan_for1_2611533029
</div>

```python
# Buat file dengan nama perulangan_for1_NIM.py
# Buat program untuk perulangan for dalam pytohn
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_3029 = int(input("Masukkan jumlah perulangan: "))

for i_3029 in range(ulang_3029):
    print(f"Perulangan ke-{i_3029}")
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

Penjelasan:

2. `for i_3029 in range(ulang_3029):` digunakan untuk melakukan perulangan sebanyak jumlah yang telah dimasukkan oleh pengguna. Fungsi `range()` menghasilkan urutan angka yang dimulai dari `0` hingga sebelum nilai `ulang_3029`.
3. `print(f"Perulangan ke-{i_3029}")` digunakan untuk menampilkan nomor perulangan pada layar. Huruf `f` pada f-string memungkinkan nilai dari variabel `i_3029` dimasukkan langsung ke dalam teks.
4. Variabel `ulang_3029` digunakan untuk menyimpan jumlah perulangan, sedangkan `i_3029` digunakan sebagai variabel penghitung pada setiap proses perulangan.

### 2.2.2 Program perulangan_for2_2611533029
</div>

```python
# Buat file dengan nama perulangan_for2_NIM.py
# Buat program untuk perulangan for dalam pytohn
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_3029 = int(input("Masukkan jumlah perulangan: "))
print("perulangan ke-0 sampai ke-",ulang_3029-1)
for i_3029 in range(ulang_3029):
    print(i_3029,end=" ")
print()
print("perulangan ke-1 sampai ke-",ulang_3029)
for i_3029 in range(ulang_3029):
    print(i_3029+1,end=" ")
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

Penjelasan:
1. `ulang_3029 = int(input("Masukkan jumlah perulangan: "))` digunakan untuk menerima input jumlah perulangan dari pengguna dan mengubahnya menjadi tipe data integer.
2. `print("perulangan ke-0 sampai ke-", ulang_3029-1)` digunakan untuk menampilkan informasi bahwa perulangan pertama akan menghasilkan angka dari `0` sampai satu angka sebelum nilai `ulang_3029`.
3. `for i_3029 in range(ulang_3029):` digunakan untuk melakukan perulangan berdasarkan nilai yang dimasukkan. Fungsi `range()` menghasilkan angka mulai dari `0` hingga `ulang_3029-1`.
4. `print(i_3029, end=" ")` digunakan untuk menampilkan nilai `i_3029` pada setiap perulangan. Parameter `end=" "` membuat hasil ditampilkan dalam satu baris dan dipisahkan dengan spasi.
5. `print()` digunakan untuk membuat baris baru setelah perulangan pertama selesai.
6. `print("perulangan ke-1 sampai ke-", ulang_3029)` digunakan untuk menampilkan informasi bahwa perulangan berikutnya akan menghasilkan angka mulai dari `1` hingga nilai `ulang_3029`.
7. `for i_3029 in range(ulang_3029):` kembali digunakan untuk melakukan perulangan sebanyak nilai `ulang_3029`.
8. `print(i_3029+1, end=" ")` digunakan untuk menampilkan nilai penghitung yang telah ditambah `1`, sehingga angka yang ditampilkan dimulai dari `1` sampai `ulang_3029`.

### 2.2.3 Program perulangan_for3_2611533029
</div>

```python
# Buat file dengan nama perulangan_for3_NIM.py
# Buat program untuk perulangan for dalam pytohn
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_3029 = int(input("Masukkan jumlah perulangan: "))

jumlah_3029 = 0
for i_3029 in range(1,ulang_3029+1):
    print(i_3029,end=" ")
    jumlah_3029 = jumlah_3029 + i_3029

    if i_3029 < ulang_3029:
        print(" + ",end=" ")
    else:
        print(" = ",jumlah_3029,end=" ")
print()
print("Jumlah =",jumlah_3029)
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

Penjelasan:
1. `ulang_3029 = int(input("Masukkan jumlah perulangan: "))` digunakan untuk menerima jumlah perulangan dari pengguna dan mengubah input menjadi tipe data integer.
2. `jumlah_3029 = 0` digunakan untuk membuat variabel penampung yang menyimpan hasil penjumlahan. Nilai awalnya ditetapkan `0`.
3. `for i_3029 in range(1, ulang_3029+1):` digunakan untuk melakukan perulangan mulai dari angka `1` hingga nilai `ulang_3029`. Penambahan `1` pada batas akhir `range()` dilakukan agar nilai `ulang_3029` ikut diproses.
4. `print(i_3029, end=" ")` digunakan untuk menampilkan angka pada setiap perulangan dalam satu baris. Parameter `end=" "` memberikan spasi setelah setiap angka.
5. `jumlah_3029 = jumlah_3029 + i_3029` digunakan untuk menambahkan nilai `i_3029` ke dalam variabel `jumlah_3029` sehingga setiap angka yang muncul akan dijumlahkan.
6. `if i_3029 < ulang_3029:` digunakan untuk memeriksa apakah angka yang sedang diproses masih lebih kecil dari jumlah perulangan.
7. `print(" + ", end=" ")` digunakan untuk menampilkan tanda `+` di antara angka-angka yang sedang dijumlahkan.
8. `else:` dijalankan ketika nilai `i_3029` sudah mencapai angka terakhir dalam perulangan.
9. `print(" = ", jumlah_3029, end=" ")` digunakan untuk menampilkan tanda `=` beserta hasil penjumlahan seluruh angka.
10. `print()` digunakan untuk berpindah ke baris baru setelah proses perulangan selesai.
11. `print("Jumlah =", jumlah_3029)` digunakan untuk menampilkan hasil akhir penjumlahan yang telah disimpan dalam variabel `jumlah_3029`.


### 2.2.4 Program nested_for1_2611533029
</div>

```python
# Buat file dengan nama nested_for1_NIM.py
# Buat program untuk perulangan for dalam pytohn
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

batas_3029 = int(input("Masukkan nilai batas: "))
for line_3029 in range(1,batas_3029 + 1):
    for j_3029 in range(1,(-1 * line_3029 + batas_3029) + 1):
        print(".",end="")
    print(line_3029)
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

Penjelasan:

2. `for line_3029 in range(1, batas_3029 + 1):` digunakan sebagai perulangan bagian luar yang berjalan mulai dari angka `1` hingga nilai `batas_3029`.
3. `for j_3029 in range(1, (-1 * line_3029 + batas_3029) + 1):` digunakan sebagai perulangan bagian dalam untuk menentukan jumlah karakter titik (`.`) yang akan dicetak. Jumlah titik akan berkurang ketika nilai `line_3029` bertambah.
4. `print(".", end="")` digunakan untuk mencetak karakter titik tanpa berpindah ke baris baru. Parameter `end=""` membuat titik-titik dicetak secara berurutan.
5. `print(line_3029)` digunakan untuk menampilkan nilai `line_3029` setelah seluruh titik pada baris tersebut selesai dicetak.

### 2.2.7 Program nested_for4_2611533029
</div>

```python
# Buat file dengan nama nested_for4_NIM.py
# Buat program untuk perulangan for dalam pytohn
# Nama variabel ditambah 4 digit nim terakhir c_3029ontoh: ulang_1234
# Program ini menggunakan fungsi input()

tinggi_3029 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_3029 % 2 != 0:
    print("Tinggi pola harus bilangan genap.")
else:
    a_3029 = tinggi_3029
    c_3029 = a_3029
    lebar_3029 = (2 * tinggi_3029) - 2

    for i_3029 in range(1, tinggi_3029 + 1):
        b_3029 = c_3029 + 1

        for j_3029 in range(1, lebar_3029 + 1):

            # Baris atas dan bawah
            if i_3029 == 1 or i_3029 == tinggi_3029:
                if j_3029 == 1 or j_3029 == lebar_3029:
                    print("#", end="")
                else:
                    print("*", end="")
            # Baris isi
            else:
                if j_3029 == 1 or j_3029 == lebar_3029:
                    print("|", end="")
                else:
                    if j_3029 == c_3029:
                        print("<", end="")
                    elif j_3029 == b_3029:
                        print(">", end="")
                    elif j_3029 == (lebar_3029 - c_3029):
                        print("<", end="")
                    elif j_3029 == (lebar_3029 - c_3029 + 1):
                        print(">", end="")
                    elif j_3029 > b_3029 and j_3029 < (lebar_3029 - c_3029):
                        print(".", end="")
                    else:
                        print(" ", end="")

        print()

        # logika asli java
        a_3029 -= 2
        if a_3029 <= 0:
            c_3029 = -a_3029 + 2
        else:
            c_3029 = a_3029
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

Penjelasan:
1. `tinggi_3029 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))` digunakan untuk meminta pengguna memasukkan tinggi pola dan mengubah input menjadi tipe data integer.

3. `print("Tinggi pola harus bilangan genap.")` digunakan untuk menampilkan pesan apabila pengguna memasukkan nilai tinggi yang bukan bilangan genap.
4. `else:` digunakan untuk menjalankan kode pembuatan pola apabila nilai `tinggi_3029` merupakan bilangan genap.
5. `a_3029 = tinggi_3029` dan `c_3029 = a_3029` digunakan untuk menyimpan nilai awal yang nantinya digunakan dalam menentukan posisi karakter pada pola.
6. `lebar_3029 = (2 * tinggi_3029) - 2` digunakan untuk menghitung lebar pola berdasarkan nilai tinggi yang dimasukkan.
7. `for i_3029 in range(1, tinggi_3029 + 1):` digunakan untuk melakukan perulangan pada setiap baris pola, mulai dari baris pertama hingga baris terakhir.
8. `b_3029 = c_3029 + 1` digunakan untuk menentukan posisi karakter `>` berdasarkan nilai `c_3029`.
9. `for j_3029 in range(1, lebar_3029 + 1):` digunakan untuk melakukan perulangan pada setiap posisi kolom dalam satu baris pola.
10. `if i_3029 == 1 or i_3029 == tinggi_3029:` digunakan untuk memeriksa apakah program sedang memproses baris paling atas atau paling bawah.
11. `if j_3029 == 1 or j_3029 == lebar_3029:` digunakan untuk menentukan posisi sudut kiri dan kanan pada baris atas atau bawah. Pada posisi tersebut program mencetak karakter `#`.
12. `print("*", end="")` digunakan untuk mencetak karakter `*` pada bagian selain sudut di baris paling atas dan bawah.
13. `if j_3029 == 1 or j_3029 == lebar_3029:` pada bagian baris isi digunakan untuk mencetak karakter `|` pada sisi kiri dan kanan pola.
14. `if j_3029 == c_3029:` digunakan untuk menentukan posisi karakter `<` pada bagian dalam pola.
15. `elif j_3029 == b_3029:` digunakan untuk menentukan posisi karakter `>` berdasarkan nilai `b_3029`.
16. `elif j_3029 == (lebar_3029 - c_3029):` dan `elif j_3029 == (lebar_3029 - c_3029 + 1):` digunakan untuk menentukan posisi karakter `<` dan `>` pada bagian sisi pola yang berlawanan.
17. `elif j_3029 > b_3029 and j_3029 < (lebar_3029 - c_3029):` digunakan untuk menentukan bagian ruang di antara posisi karakter tertentu. Pada kondisi ini program mencetak karakter `.`.
18. `print(" ", end="")` digunakan untuk mencetak spasi pada posisi yang tidak memenuhi kondisi sebelumnya sehingga pola dapat terbentuk dengan posisi yang sesuai.

```python
tinggi_3029 = int(input("Masukkan tinggi segitiga: "))

for i in range(1, tinggi_3029 + 1):
    print(" " * (tinggi_3029 - i), end="")  # cetak spasi
    for j in range(1,i + 1):
        print("*", end=" ")
    print()  # pindah ke baris berikutnya
```
